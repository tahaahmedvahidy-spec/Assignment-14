```python
# Step 1: Import the libraries needed to analyze data and draw charts.
# pandas: works with tables/dataframes.
# matplotlib and seaborn: draw graphs.
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

# Step 2: Apply a readable, beginner-friendly chart style.
sns.set_theme(style="whitegrid")
print("Libraries imported successfully!")
```

```python
#Dataset 1 — Pakistani E-Commerce Data
```

```python
# Step 3: Locate and load the Pakistani E-Commerce CSV file.
# The CSV must be in the same folder as the Jupyter notebook.
ecommerce = pd.read_csv(r"C:\Users\dell\Downloads\pakistani_ecommerce_dataset.csv")

# Step 4: Check the number of rows/columns and preview the first five records.
print("E-commerce dataset shape (rows, columns):", ecommerce.shape)
display(ecommerce.head())
print("Missing values per column:\n", ecommerce.isnull().sum())
```

```python
# Dataset 1 — Q1: Find the most popular product category.
# Step 5: Count how often each product category appears.
category_counts = ecommerce["product_category"].value_counts()
most_popular_category = category_counts.idxmax()
most_popular_category_count = category_counts.max()

print("Q1. Most popular product category:", most_popular_category)
print("Number of orders:", most_popular_category_count)
display(category_counts.rename("Number of orders").to_frame())
```

```python
# Dataset 1 — Q2: Calculate the average order value.
# Step 6: Use the mean() function on the order_value column.
average_order_value = ecommerce["order_value"].mean()
print(f"Q2. Average order value: {average_order_value:,.2f} (currency not specified in CSV)")
```

```python
# Dataset 1 — Q3: Find the city with the highest number of orders.
# Step 7: Count the occurrences of every city.
city_order_counts = ecommerce["city"].value_counts()
city_with_most_orders = city_order_counts.idxmax()
most_orders = city_order_counts.max()

print("Q3. City with the highest number of orders:", city_with_most_orders)
print("Number of orders:", most_orders)
display(city_order_counts.rename("Number of orders").to_frame())
```

```python
# Dataset 1 — Q4: Identify the most used payment method.
# Step 8: Count each payment method and select the most frequent one.
payment_counts = ecommerce["payment_method"].value_counts()
most_used_payment_method = payment_counts.idxmax()
most_used_payment_count = payment_counts.max()

print("Q4. Most used payment method:", most_used_payment_method)
print("Number of orders:", most_used_payment_count)
display(payment_counts.rename("Number of orders").to_frame())
```

```python
# Dataset 1 — Q5: Draw a bar chart of average order value by city.
# Step 9: Group all orders by city, then find each city's average order value.
city_average_values = (
    ecommerce.groupby("city", as_index=False)["order_value"]
    .mean()
    .sort_values("order_value", ascending=False)
)
print("Q5. Average order value by city (currency not specified):")
display(city_average_values.round(2))

# Step 10: Create, label and display the required bar chart.
plt.figure(figsize=(9, 5))
ax = sns.barplot(data=city_average_values, x="city", y="order_value", color="tab:blue")
ax.set(title="Average Order Value by City", xlabel="City", ylabel="Average Order Value (dataset units)")
ax.bar_label(ax.containers[0], fmt="%.0f", padding=3)
ax.set_ylim(0, city_average_values["order_value"].max() * 1.13)
plt.tight_layout()
plt.show()
```

```python
#Dataset 2 — Student Lifestyle & Study Habits
```

```python
# Step 11: Locate and load the Student Lifestyle CSV file.
students = pd.read_csv(r"C:\Users\dell\Downloads\student_lifestyle_dataset.csv")

# Step 12: Preview the data and check for missing values.
print("Student lifestyle dataset shape (rows, columns):", students.shape)
display(students.head())
print("Missing values per column:\n", students.isnull().sum())
```

```python
# Dataset 2 — Q1: Calculate the average number of study hours.
# Step 13: Find the mean of the study_hours column.
average_study_hours = students["study_hours"].mean()
print(f"Q1. Average study hours: {average_study_hours:.2f} hours")
```

```python
# Dataset 2 — Q2: Compare average sleep hours for males and females.
# Step 14: Group students by gender and calculate their mean sleep hours.
sleep_by_gender = students.groupby("gender")["sleep_hours"].mean()
print("Q2. Average sleep hours by gender:")
display(sleep_by_gender.round(2).rename("Average sleep hours").to_frame())
print("Gender sleeping more on average:", sleep_by_gender.idxmax())
```

```python
# Dataset 2 — Q3: Check whether screen time and sleep hours are related.
# Step 15: Calculate Pearson correlation (r), which ranges from -1 to +1.
# r near 0 = almost no linear relationship; r near +/-1 = strong relationship.
screen_sleep_correlation = students["screen_time"].corr(students["sleep_hours"])
print(f"Q3. Pearson correlation between screen time and sleep: {screen_sleep_correlation:.4f}")
if abs(screen_sleep_correlation) < 0.10:
    print("Interpretation: Virtually no linear relationship in this dataset.")
elif screen_sleep_correlation > 0:
    print("Interpretation: A positive linear relationship.")
else:
    print("Interpretation: A negative linear relationship.")

# Step 16 (extra): Use a scatter plot to visually inspect the relationship.
plt.figure(figsize=(7, 5))
sns.scatterplot(data=students, x="screen_time", y="sleep_hours", color="tab:blue")
plt.title("Screen Time vs Sleep Hours")
plt.xlabel("Screen Time (hours)")
plt.ylabel("Sleep Hours (hours)")
plt.tight_layout()
plt.show()
```

```python
# Dataset 2 — Q4: Draw a histogram of students' daily transport cost.
# Step 17: A histogram shows how transport costs are distributed.
plt.figure(figsize=(8, 5))
sns.histplot(data=students, x="daily_transport_cost", bins=10, edgecolor="white", color="tab:blue")
plt.title("Distribution of Daily Transport Cost")
plt.xlabel("Daily Transport Cost (dataset units)")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
print("Q4. Histogram of daily transport cost displayed above.")
```

```python
# Dataset 2 — Q5: Compare average internet usage for males and females.
# Step 18: Group by gender and find the mean internet usage.
internet_by_gender = (
    students.groupby("gender", as_index=False)["internet_usage"]
    .mean()
    .sort_values("gender")
)
print("Q5. Average internet usage by gender (dataset units):")
display(internet_by_gender.round(2))

# Step 19 (extra): Draw a bar chart to make the comparison clearer.
plt.figure(figsize=(7, 5))
ax = sns.barplot(data=internet_by_gender, x="gender", y="internet_usage", color="tab:blue")
ax.set(title="Average Internet Usage by Gender", xlabel="Gender", ylabel="Average Internet Usage (dataset units)")
ax.bar_label(ax.containers[0], fmt="%.2f", padding=3)
ax.set_ylim(0, internet_by_gender["internet_usage"].max() * 1.15)
plt.tight_layout()
plt.show()
```
